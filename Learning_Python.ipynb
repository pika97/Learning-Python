```python
import numpy as np, datetime
def grade(topic, task, got, expected, out_of=10):
    ok = np.allclose(got, expected)
    score = out_of if ok else 0
    print("✅ Correct" if ok else "❌ Not yet", "-", task)
    print(f"TRK|{topic}|{task}|{score}|{datetime.date.today()}")

# example:
# grade("numpy", "Reshape", arr.reshape(2,3), [[1,2,3],[4,5,6]])
```

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])
answer = arr.reshape(2, 3)

grade("numpy", "Reshaping", answer, [[1, 2, 3], [4, 5, 6]])
```

Feauture Normalization or Normalization is basically rescaling your numbers so they are measured in how far they are from average instead of in their orignal units.

```python
import numpy as np
x = np.array([
     [1, 2, 3],
     [2, 3, 4],
     [5, 6, 7],
     [1, 2, 3]
])
mean = x.mean(axis=0)
std = x.std(axis=0)
x_scaled = (x - mean)/std                      # Normalization
print(x_scaled)
```

```python
import numpy as np
rng = np.random.default_rng(42)

print(rng.random(4))                       # Random decimal numbers from 0 to 1


print(rng.integers(1, 10, size=5))         # Random integers from 1 to 10 (10 excluded)


print(rng.normal(0, 1, size=5))            # Random values from a normal distribution


values = np.array([10, 20, 30, 40, 50])    # Shuffle an array in place
rng.shuffle(values)
print(values)
```

```python
import numpy as np
rng = np.random.default_rng(42)
X = rng.normal(50, 10, size=(3,4))
mean = X.mean(axis=0)
std = X.std(axis=0)
X_scaled = (X - mean)/std
print("Orginal Shape", X.shape)
print("X_scaled Shape", X_scaled.shape)
print("X_scaled mean", X_scaled.mean(axis=0))
print("X_scaled Std", X_scaled.std(axis=0))


```

Why normalization matters - Some Machine learning algorithms may be affected by the difference in feature scale. We often therefore scale numerical features before training.

Imagine two features
1. Age: 25-55
2. Cholestrol: 180-240

Two Common Methods
1. Min-Max Normalization
   X_scaled = X - X_min/ X_max - X_min

   The smallest value becomes 0 and the largest becomes 1.
        

```python
values = np.array([10, 20, 30, 40, 50])
normalized = (values - values.min())/(values.max()  - values.min())
print(normalized)
```

2. Standardization (Z-score scaling): Instead of forcing values from 0 to 1 range, It transforms them so that each feature has.
   Mean approximately 0
   Standard deviation approximately 1

   Z = (X - miu)/ sigma

   X = orignal value
   miu = mean
   sigma = standard deviation

```python
import numpy as np
values = np.array([10, 20, 30, 40, 50])
mean = values.mean()
std = values.std()

Z = (values - mean)/std
print(Z)
```

A value of zero is exactly at the average. A positive values above the average and negative below the average.

```python
import numpy as np
p_data = np.array([
          [29, 10, 78],
          [68, 87, 120],
          [37, 76, 87],
          [36, np.nan, 98],
          [68, 46, 68],
          [37, 39, np.nan],
          [25, 81, np.nan],
          [68, np.nan, 67],
          [78, 19, 86],
          [36, 49, 97]
])
print("Missing values count:", np.isnan(p_data).sum())

result = np.array([0, 1, 1, 1, 0, 0, 1, 0, 0, 0])

# Splitting into training and testing indices first to avoid data leakage
rng = np.random.default_rng(42)
indices = rng.permutation(len(p_data))

split = int(0.8 * len(p_data))
train_idx = indices[:split]
test_idx = indices[split:]

ptrain = p_data[train_idx]
ptest = p_data[test_idx]

result_train = result[train_idx]
result_test = result[test_idx]

# Imputing missing values using training column means only
column_mean = np.nanmean(ptrain, axis=0)

rows_tr, cols_tr = np.where(np.isnan(ptrain))
ptrain[rows_tr, cols_tr] = column_mean[cols_tr]

rows_ts, cols_ts = np.where(np.isnan(ptest))
ptest[rows_ts, cols_ts] = column_mean[cols_ts]

# Calculate statistics from training data
mean = ptrain.mean(axis=0)
std = ptrain.std(axis=0)

Scaled_ptrain = (ptrain - mean) / std
Scaled_ptest = (ptest - mean) / std

print("\nTraining Data (Scaled):\n", Scaled_ptrain)
print("\nTesting Data (Scaled):\n", Scaled_ptest)
```

Outlier Analysis

An outlier is not automatically an error. It may represent a genuine and important event.

Q1 = 25th percentile.
Q3 = 75th percentile.
IQR = the middle 50% spread of the data.

```python
import numpy as np
import pandas as pd
closure = np.array([10, 19, 12, 13, 14, 80])
print(np.mean(closure))

# calculate quartiles

Q1 = np.percentile(closure, 25)
Q3 = np.percentile(closure, 75)

IQR = Q3 - Q1
print(IQR)

# calculate outlier boundaries

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit", lower_limit)
print("Upper Limit", upper_limit)

# identify outliers

Outliers = closure[(closure<lower_limit) | (closure>upper_limit)]  # (|) is an OR symbol
print("Outliers", Outliers)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
                "Training Hours": [3, 5, 2, 7, 2, 8],
                "Manufacturing Errors": [0, 4, 1, 6, 2, 3],
                "Closure Days": [8, 5, 7, 4, 8, 3]
                })
correlation = df.corr()
print(correlation)

# to visualize correlation matrix use seaborn

import seaborn as sns
sns.heatmap(correlation, annot=True, cmap="coolwarm")

plt.title("Quality Data Correlation Heatmap")
plt.show()
```


- Values close to +1: strong positive relationship.
- Values close to -1: strong negative relationship.
- Values close to 0: weak linear relationship.
The diagonal is always 1.0, because every variable has a perfect correlation with itself.

AI engineering connection: Heatmaps can help you identify potentially redundant features before training a model. However, don't automatically remove highly correlated features; consider the model, domain knowledge and validation results.

If the model performs well on the training data but not on the testing data then the model is "overfitting".

Overfitting mean model has memorized patterns in the training data rathar than learning patterns that generalize well to new data.

1. What is data leakage?
Data leakage happens when a machine-learning model gets access to information during training that it would not legitimately have when making a real-world prediction.

Examples of Data Leakage

When training a model, you do provide the correct answers (y) separately so the model can learn from them.

But you must not include those answers inside the input features (X).

DON'T COMPUTE MEAN AND STD FOR TESTING DATA SEPERATELY.

split the data first
compute the parameters like mean and std using training data
Scale the training data
use the parameters from training data to scale the testing data

Feauture Scaling - Imaging the pharma Quality Incident data set has two features

Investigation Complexity : 1 - 10
Closure                  : 5 - 100

These values are on different scales. Some ML algorithms can be affected by these differences. Feature Scaling puts these numerical features on to comparable scales.


There are two techiques to do that
1. Min-Max Normalization (Converts values in range 0-1)
2. Standardization( Converts values based on mean and std)

Min-Max Normalization

X_scaled = X - Xmin / Xmax - Xmim

Normalization

Z = (X - miu) / std

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

df = pd.DataFrame({
                  "Investigation Complexity": [2, 4, 8, 2, 7, 5, 3],
                  "Closure": [10, 20, 30, 40, 50, 60, 70]
})

# Min-Max Scaling
minmax = MinMaxScaler()
normalized_data = minmax.fit_transform(df)
print("Min-Max Normalized Data:")
print(normalized_data)

# Standardized Scaling
standard = StandardScaler()
standardized_data = standard.fit_transform(df)
print("Standardized Data:")
print(standardized_data)
```
