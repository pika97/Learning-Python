A mini Machine Learning data preparation workflow


a. Create a test database (may be add missing values and fix them as well)

b. Split into training and testing data

c. Calculate training values from training data d. standardize both datasets using those training values.

```python
import numpy as np
p_data = np.array([
          [29, 10, 78],
          [68, 87, 120],
          [37, 76, 87],
          [36, np.nan, 98],
          [68, 46, 68],
          [37, 39, np.nan],
          [25, 81, np.nan],
          [68, np.nan, 67],
          [78, 19, 86],
          [36, 49, 97]

])
print(np.isnan(p_data).sum())

result = np.array([0, 1, 1, 1, 0, 0, 1, 0, 0, 0])

# Dividing into training and testing data
rng = np.random.default_rng(42)
indices = rng.permutation(len(p_data))

split = int(0.8*len(p_data))

train_idx = indices[:split]
test_idx = indices[split:]

ptrain = p_data[train_idx]
ptest = p_data[test_idx]

result_train = result[train_idx]
result_test = result[test_idx]

# Now we need the missing numbers by "column means" (from TRAINING data only)

column_mean = np.nanmean(ptrain, axis=0)         # means from ptrain only

rows, columns = np.where(np.isnan(ptrain))       # locating the coordinates of missing values
ptrain[rows, columns] = column_mean[columns]

rows, columns = np.where(np.isnan(ptest))        # fill test gaps with the TRAIN means
ptest[rows, columns] = column_mean[columns]

# calculate statistics from training data

mean = ptrain.mean(axis=0)
std = ptrain.std(axis=0)

Scaled_ptrain = (ptrain - mean)/std
Scaled_ptest = (ptest - mean)/std

print("Training Data", Scaled_ptrain)
print("Testing Data", Scaled_ptest)
```
