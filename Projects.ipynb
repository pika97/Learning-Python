A mini Machine Learning data preparation workflow


a. Create a test database (may be add missing values and fix them as well)

b. Split into training and testing data

c. Calculate training values from training data d. standardize both datasets using those training values.

```python
import numpy as np
p_data = np.array([
          [29, 10, 78],
          [68, 87, 120],
          [37, 76, 87],
          [36, np.nan, 98],
          [68, 46, 68],
          [37, 39, np.nan],
          [25, 81, np.nan],
          [68, np.nan, 67],
          [78, 19, 86],
          [36, 49, 97]

])
print(np.isnan(p_data).sum())

result = np.array([0, 1, 1, 1, 0, 0, 1, 0, 0, 0])

# Dividing into training and testing data
rng = np.random.default_rng(42)
indices = rng.permutation(len(p_data))

split = int(0.8*len(p_data))

train_idx = indices[:split]
test_idx = indices[split:]

ptrain = p_data[train_idx]
ptest = p_data[test_idx]

result_train = result[train_idx]
result_test = result[test_idx]

# Now we need the missing numbers by "column means" (from TRAINING data only)

column_mean = np.nanmean(ptrain, axis=0)         # means from ptrain only

rows, columns = np.where(np.isnan(ptrain))       # locating the coordinates of missing values
ptrain[rows, columns] = column_mean[columns]

rows, columns = np.where(np.isnan(ptest))        # fill test gaps with the TRAIN means
ptest[rows, columns] = column_mean[columns]

# calculate statistics from training data

mean = ptrain.mean(axis=0)
std = ptrain.std(axis=0)

Scaled_ptrain = (ptrain - mean)/std
Scaled_ptest = (ptest - mean)/std

print("Training Data", Scaled_ptrain)
print("Testing Data", Scaled_ptest)
```

Mini Project: Prepare Healthcare Data

Your task is to create a dataset, inspect it, identify an outlier and standardise the features.


```python
import numpy as np
p_data = np.array([
                   [25, 120, 180],
                   [30, 125, 190],
                   [35, 130, 200],
                   [40, 135, 210],
                   [45, 140, 220],
                   [50, 145, 230],
                   [55, 150, 240],
                   [60, 155, 500]
])

# dividing the data in 1st quarter and 3rd quarter

Q1 = np.percentile(p_data, 25)
Q3 = np.percentile(p_data, 75)
IQR = Q3 - Q1

Lower = Q1 - 1.5*IQR
Upper = Q3 + 1.5*IQR

Outlier = p_data[(p_data<Lower)|(p_data>Upper)]
print("Outlier", Outlier)

mean = p_data.mean(axis=0)
std = p_data.std(axis=0)

Scaled_p_data = (p_data - mean)/std
print(Scaled_p_data)
```

Mini Project - Clean and analyze a quality dataset.

Your task is to:

1.Create a DataFrame.
2.Inspect the records.
3.Find missing values.
4.Fill missing investigation durations.
5.Filter high-priority incidents.
6.Calculate average closure time by department

```python
import pandas as pd
import numpy as np

# creating a Quality Data Set

data = {
    "Department" : [
        "Manufacturing",
        "QC",
        "Manufacturing",
        "QC",
        "Engineering",
        "Quality Assurance",
        "Manufacturing",
        "Quality Assurance",
        "Quality Assurance",
        "Manufacturing",
        "Engineering",
        "Quality Assurance",
        "Quality Assurance",
        "Quality Assurance",
        "Engineering"
    ],
    "Priority": [
                "High",
                "High",
                "Low",
                "High",
                "Low",
                "Low",
                "High",
                "High",
                "Low",
                "Low",
                "High",
                "Low",
                "High",
                "High",
                "High"
    ],
    "Closure": [12, 9, 12, 15, np.nan, 5, 14, 11, np.nan, 9, 11, 12, 18, 10, 7]
    }

df = pd.DataFrame(data)

# inspect data

print(df.head())
print(df.info())

# find missing values

print(df.isnull().sum())

# fill the missing values with mean

mean_closure = df["Closure"].mean()
df["Closure"] = df["Closure"].fillna(mean_closure)


# filter high priority incidents

high_priority = df[df["Priority"] == "High"]
print(high_priority)

# grouping and analzing
# total number of closure per department

result = df.groupby("Department")["Closure"].sum()
print(result)

# average closure by department

avg_closure = df.groupby("Department")["Closure"].mean()
print(avg_closure)

# save cleaned dataset

df.to_csv("Cleaned_qality_data.csv", index=False)
```
