```python
def f(y, x, z):
    if x>y:
        return x+y
        if x<y:
            return y+z
        else:
            return z+x
    else:
      return x+y+z
print(f(6,4,8))

# after the 1st if condition is false skips the second if (dead code) and jumps to the "else"
```
