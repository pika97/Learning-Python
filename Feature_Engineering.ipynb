FEATURE ENGINEERING

The process of selecting, creating or modifying features like input variable pr data to help ML models to learn patterns more effectively. It involves transforming raw data into meaningful inputs that improves models accuracy and performance.

The process involves:
1. Feature Creation
2. Feature Transformation
3. Feature Extraction
4. Feature Selection
5. Feature Scaling

Common Techniques in Feature Engineering is One Hot Encoding.


One Hot Encoding - Converts categorical variables into binary indicators, allowing them to be used by machine learning models.

```python
import pandas as pd
data = {'Color':['Red', 'Blue', 'Yellow', 'Blue']}                 # create a dictionary key 'Colors'
df = pd.DataFrame(data)
df_encoded = pd.get_dummies(df, columns=['Color'], prefix='Color') # pd.dummies converts categorical variables into binary indicators, column=['Color'] specifies which column to encode, prefix='Color' prepends the name of the original column to the new column headers, resulting in the new columns: Color_Blue, Color_Red, and Color_Yellow.

print(df_encoded)
```
