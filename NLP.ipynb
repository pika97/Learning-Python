Bag of Word - Counts how many times each word appeared in the docs

```python
documents = [
            "Incorrect reagent label",
            "Equipment failure during testing",
            "Incorrect label during packaging"
]
```

creates numerical representation of each sentence
It then creates a numerical representation of each sentence.

A value of 1 means the word appears; 0 means it does not.

Implement NLP in python. We will use count vectorizer from Scikit.

```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
            "Incorrect label on reagent pack",
            "Equipment failed during testing",
            "Missing documentation in batch record",
            "Incorrect reagent label on product"
]

vectorizer = CountVectorizer()                  # creates a text to number converter

X = vectorizer.fit_transform(documents)         # transforms text into numbers
print("Vocabulary:")
print(vectorizer.get_feature_names_out())       # extracts the learned vocabulary from the vectorizer
print("\nVectorized Documents:")
print(X.toarray())                              # displays the numercial representation as an array

```

Using NLP for Text Classification
Lets say we have Compalaint data with categories assigned
1. Incorect reagent label - labelling issue
2. Equipment breakdown  - Equipment issue
3. Adverse readings  -  OOS issue
4. Missing operator signs - Training issue
5. Missing batch records -  Documentation issue

Now the above has an input and defined labelled category so this is supervised learning.

We could train a model to learn relation between text and the category

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline

text_document = [
                "Incorrect reagent label",
                "Missing date in batch records",
                "Equipment failure",
                "Defective raw material",
                "Missing operators sign",
                "Equipment not calibrated"
]

labels = ["Labelling issue",
          "Documentation issue",
          "Equipment issue",
          "Supplier issue",
          "Training issue",
          "Equipment issue"
          ]

model = make_pipeline(TfidfVectorizer(),
                      LogisticRegression(max_iter=1000)
          )

model.fit(text_document, labels)

new_complaint = ["wrong date on the batch manufacuring records"]

prediction = model.predict(new_complaint)
print(prediction)
```
