1. Tokenization
Splitting sentenses into smaller units called tokens usually worlds.

```python
sentence = "Incorrect reagent label found"
tokens = sentence.split()

print(tokens)
```

Stop words
The, is, a, an, and, of

" The reagent label is incorrect "

After removing selected stop words:
['reagent', 'label', 'incorrect']

Important: Stop words should not always be removed. Words such as "not" and "no" can completely change the meaning of a complaint. For example, "not defective" is very different from "defective"
