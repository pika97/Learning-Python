Whats a neural network?
1. A neural network is inspired by the way humans process information.
2. It consists of interconnected units called neurons which process information and learn patterns from data.

Structure of a Neural Network has 3 main layers.
1. Input layer
Recieves the input data
2. Hidden Layer
This is where the pattern learns relationship and patterns. A network will have one or more hidden layers, each layer processes information and pass on to next layer.
3. Output Layer
Produces final prediction.

Whats a neuron?
A neuron recieves input, assigns weight to them, adds a bias and passess the result through activation function.

Weight - Determines how much influence each input has on the prediction.

    Input          Weight
1. Complexity       0.8
2. Priority         0.7
3. Impact           0.9  

Bias - Additional adjustable value that helps the neuron shift its value

Activation Function- Tranforms the neurons calculated value. It helps neural network learn complex, non-linear relationships.

A common activation function is ReLU (Rectified Linear Unit)

ReLU(X) = max(0,X)
1. ReLU(-4) = 0
2. ReLU(4) = 4
3. ReLU(0) = 0

Another activation function is Sigmoid. It converts values into number between 0 and 1, useful for binary outputs.

How does a Neural Network learn?
1. Input data - Providing Training Examples
2. Forward Propogation - Calculate Prediction
3. Calculate Loss - Measure Prediction Error
4. Back Propogation - Contribute how weight contribute to error.
5. Weight Update - Adjust weight to reduce the loss.
6. Repeat the training process.

1. Epochs - One complete pass through the entire dataset.
2. Batch Size - Number of Training Examples processed before an update.
3. Learning rate - Controls how large the weights updates are.

Suppose if you have
1. 1000 training records
2. Epochs = 10
3. Batch size = 100

Therefore 1000/100 = 10 batches per epoch. So 100 batches across 10 epochs.

What happens if the learning rate is too high?

The model may make excessively large updates and fail to settle on good weights.

What happens if it is too low?

Training may take a very long time because the model makes very small updates.

Building a Neural Network using MLP Classifier (Multi Layer Perceptron Classifier.

```python
# ARTIFICIAL NEURAL NETWORK ANN/MLP

from sklearn.neural_network import MLPClassifier

# Training data

X = [
    [1, 0, 3],
    [2, 4, 2],
    [0, 0, 3],
    [1, 1, 0],
    [3, 2, 1],
    [2, 2, 1]
]

# Target data
y = ["No", "Yes", "Yes", "No", "No", "No"]

# create a neural network
model = MLPClassifier(
                     hidden_layer_sizes=(5,),
                     activation="relu",
                     max_iter=2000,
                     random_state=42
)


# hidden_layer_sizes=(5,) is 5 nuerons in one hidden layer
# activation="relu" is using relu as activation function
# max_iter=2000 allows upto 2000 training iterations


# train model
model.fit(X,y)

predict_new_data = [[2, 2, 2]]

prediction = model.predict(predict_new_data)

print("Prediction:", prediction)

```

Whats a Convolution Neural Network (CNN)
1. is a deep learning model to analyze visual data
2. CNNs are especially useful because they learn visual features automatically.

For example - When analyzing an image, a CNN may learn to identify
1. Simple Edges and lines
2. Shapes & textures
3. More complex stuctures
4. Higher level visual patterns

How does the CNN work?
1. Input Image - Numerical Pixel values
2. Convolution - Detects visual features
3. Activation ReLU - Introduces Non-Linearity
4. Pooling - Reduces spatial dimensions
5. Fully connected layer - combines learned features
6. Output - Predicted class or probability


Convolution is basically moving a small matrix or kernal over an image to detect specific features.
A filter is a small grid of numbers. As it moves over the image, it performs calculations on nearby pixel values.

For example, a filter can learn to detect:
- Horizontal edges
- Vertical edges
- Diagonal edges
- Textures and shapes

The resulting image-like array is called a feature map. It shows where the filter detected its learned pattern.

The filter values are generally learned during training rather than manually specified for every feature.

ReLU Activation Function - After Convolution CNN applies activation function called ReLU (Rectified Linear Unit)
1. ReLU(x) = max(0,x) It replaces positive values with zeros and negative values unchanged.

Why use it?
It introduces non-linearity, allowing the network to learn more complex patterns than a purely linear model could learn.

Pooling - Reduces spatial size of a feature map while retaining important information.
1. A common technique is Max Pooling. - Max Pooling selects the largest value in each region.

Fully Connected Layer and Output.
1. After convolution and pooling the network combines the learned features to make predictions.
2. A fully connected layer recieves information from the preceding layer and learns how those features relate to output categories.

Three important vision task
1. Image Classification - Assigns label to an entire image
Example - Classifies an X-ray as normal or abnormal.
2. Object Detection - Identifies objects and their locations using bounding
Example - Locate a suspicious region in an X-ray
3. Image Segmentation - Classifies individual region into pixels
Example - Outline a tumor region in an MRI scan


1. 2D CNNs analyse individual image slices.
2. 3D CNNs can analyse volumes, such as a complete CT or MRI scan, using depth as another spatial dimension.

Build the CNN Model

```python
import tensorflow as tf
from tensorflow.keras import models, layers
import matplotlib.pyplot as plt
import numpy as np

# load handwritten digits
(X_train, y_train), (X_test, y_test) = (tf.keras.datasets.mnist.load_data())
print(X_train.shape)
print(y_train.shape)

# normalize pixel value
X_train = X_train/255.0
X_test = X_test/255.0

# add channel dimension
X_train = X_train[..., tf.newaxis]
X_test = X_test[..., tf.newaxis]

print(X_train.shape)

# build CNN Model
model = models.Sequential([
    layers.Input(shape=(28,28,1)),
    # detect image features
    layers.Conv2D(32, (3,3), activation="relu"),    # Conv2D(32,(3,3)) is learns 32 filters each 3x3
    # reduce feature map dimension
    layers.MaxPooling2D((2,2)),                      # reduces spatial dimensiom
    # prepares features for classification
    layers.Flatten(),                                # flatten is converts maps into 1D vector
    # learns relationship bet features
    layers.Dense(64, activation="relu"),             # Dense(64) is learns relationsips between extracted features
    # predict one of the 10 digit classess
    layers.Dense(10, activation="softmax")           # Dense(10) is produces score 0-9 and softmax converts final scores into values that sum to 1.
])

# compile model
model.compile(
     optimizer="adam",
     loss="sparse_categorical_crossentropy",
     metrics=["accuracy"]
)

# train the model
history = model.fit(X_train, y_train, epochs=5, validation_split=0.1)

# evaluate model
test_loss, test_accuracy = model.evaluate(X_test, y_test)
print("Test Accuracy:", test_accuracy)

# predict a digit

image = X_test[0]       # select an image
prediction = model.predict(image[np.newaxis, ...])

predict_digit = np.argmax(prediction)

print("Predicted Digit:", predict_digit)
print("Actual Digit:", y_test[0])
```

Transfer Learning and pretrained model - You can build an Ai model to identify abonormalities in chest. One approch is to train the model from scratch to learn basic visual patterns such as edges, lines, curves and shapes more complex patterns and structures.

Learning all of this from the beginning requires substantial training data, computing power and time.

Transfer learning offers another approach: start with a model that has already learned useful visual features from a large dataset, then adapt it to your own task.

Pre-Trained models - Is neural networks whose weights are trained on existing datasets.

The common types of CNN Architecture are
1. VGG16
2. ResNet
3. Mobile Net
4. Efficient Net

Feature Reuse - CNN learns different levels of visual features as information passes through its layers. It may already have learned edges and boundaries, textures and patterns, combinations and shapes. So these can be reused instead of training a model from scratch.

Freezing a Layer - When we use a pretrained CNN we choose which layers are allowed to learn further.

1. Early convolution layers - frozen
2. Later convolution layers - frozen initially
3. New classification layers - trainable

Feature extraction V/S fine tuning.

Feature Extraction
Pretrained CNN - Frozen
1. Covolution Base - Existing base are preserved
2. New classification heads - Trainable (learns your new classes)
You freeze the pretrained layers and train the new classification head.

Fine Tuning
1. Earlier CNN layers remain frozen
2. Select later CNN layers - unfrozen can learn new weights
3. New classification head - trainable
