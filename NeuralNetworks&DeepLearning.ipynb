Whats a neural network?
1. A neural network is inspired by the way humans process information.
2. It consists of interconnected units called neurons which process information and learn patterns from data.

Structure of a Neural Network has 3 main layers.
1. Input layer
Recieves the input data
2. Hidden Layer
This is where the pattern learns relationship and patterns. A network will have one or more hidden layers, each layer processes information and pass on to next layer.
3. Output Layer
Produces final prediction.

Whats a neuron?
A neuron recieves input, assigns weight to them, adds a bias and passess the result through activation function.

Weight - Determines how much influence each input has on the prediction.

    Input          Weight
1. Complexity       0.8
2. Priority         0.7
3. Impact           0.9  

Bias - Additional adjustable value that helps the neuron shift its value

Activation Function- Tranforms the neurons calculated value. It helps neural network learn complex, non-linear relationships.

A common activation function is ReLU (Rectified Linear Unit)

ReLU(X) = max(0,X)
1. ReLU(-4) = 0
2. ReLU(4) = 4
3. ReLU(0) = 0

Another activation function is Sigmoid. It converts values into number between 0 and 1, useful for binary outputs.

How does a Neural Network learn?
1. Input data - Providing Training Examples
2. Forward Propogation - Calculate Prediction
3. Calculate Loss - Measure Prediction Error
4. Back Propogation - Contribute how weight contribute to error.
5. Weight Update - Adjust weight to reduce the loss.
6. Repeat the training process.

1. Epochs - One complete pass through the entire dataset.
2. Batch Size - Number of Training Examples processed before an update.
3. Learning rate - Controls how large the weights updates are.

Suppose if you have
1. 1000 training records
2. Epochs = 10
3. Batch size = 100

Therefore 1000/100 = 10 batches per epoch. So 100 batches across 10 epochs.

What happens if the learning rate is too high?

The model may make excessively large updates and fail to settle on good weights.

What happens if it is too low?

Training may take a very long time because the model makes very small updates.

Deep Learning - Deep Learning is special area in machine learning that uses neural network with multiple hidden layers.

Types of Neural Network
1. ANN/MLP - For structured Numerical Data and Classification.
2. CNN - Image Recognition and Computer Vision.
3. RNN - Sequential Data and time series pattern.
4. LSTM - Learning longer term patterns in sequences.
5. Transformer - Language Models and modern generative AI.
